```python
class Solution:
    def isValid(self, s: str) -> bool:
        st = []
        for ch in s:
            if ch in "{[(":
                st.append(ch)
            else:
                if len(st)==0:
                    return False
                topEle=st[-1]
                if ch == '}' and topEle =="{":
                    st.pop()
                elif ch==']' and topEle=='[':
                    st.pop()
                elif ch==')' and topEle=='(':
                    st.pop()
                else: 
                    return False
        return len(st)==0
        
```
