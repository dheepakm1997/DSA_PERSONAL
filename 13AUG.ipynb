```python

####2 SUM pROBLEM
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        ans=[-1,-1]
        for i in range(n):
            for j in range(i+1,n):
                if nums[i] + nums[j] ==target:
                    ans[0]=i
                    ans[1]=j
        return ans

```

```python
####Find Max Consecutive Ones



class Solution:
    def findMaxConsecutiveOnes(self, nums: List[int]) -> int:
        count=0
        maxi = 0
        for num in nums:
            if num == 0:
                count = 0
            else:
                count+=1
            maxi = max(maxi,count)
        return maxi

        
```

```python
####Two Sum Problem Optimized Approach-Hash Map###

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        mpp={}
        ans =[-1,-1]

        for i,num in enumerate(nums):
            x=target-num
            if x in mpp:
                ans[0]=i
                ans[1]=mpp[x]
                return ans
            mpp[num]=i
        return ans
```
