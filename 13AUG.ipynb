```python

####2 SUM pROBLEM
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        ans=[-1,-1]
        for i in range(n):
            for j in range(i+1,n):
                if nums[i] + nums[j] ==target:
                    ans[0]=i
                    ans[1]=j
        return ans

```

```python
####Find Max Consecutive Ones



class Solution:
    def findMaxConsecutiveOnes(self, nums: List[int]) -> int:
        count=0
        maxi = 0
        for num in nums:
            if num == 0:
                count = 0
            else:
                count+=1
            maxi = max(maxi,count)
        return maxi

        
```

```python
####Two Sum Problem Optimized Approach-Hash Map###

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        mpp={}
        ans =[-1,-1]

        for i,num in enumerate(nums):
            x=target-num
            if x in mpp:
                ans[0]=i
                ans[1]=mpp[x]
                return ans
            mpp[num]=i
        return ans
```

```python
####Single Number Problem

class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        mpp ={}
        for num in nums:
            if num not in mpp:
                mpp[num]=1
            else:
                mpp[num]+=1
        for num in nums:
            if mpp[num]==1:
                return num
        return -1
        
```

```python
###Majaority Element Problem-Hashmap Approach
class Solution:
 def majorityElement(self, nums: List[int]) -> int:

        mpp={}
        for num in nums:
            if num not in mpp:
                mpp[num]=1
            else:
                mpp[num]+=1
                
        max_freq=0
        max_num=-0
        for key,value in mpp.items():
            if max_freq<value:
                max_freq=value
                max_num = key
        return max_num
            
```

```python
####Majaority Element Problem-Sorting Approach

class Solution:
 def majorityElement(self, nums: List[int]) -> int:
    n = len(nums)
    nums.sort()

    result = nums[n//2]
    return result
```

```python
####Majority Element Problem-Boyer-Moore Voting Algorithm


class Solution:
 def majorityElement(self, nums: List[int]) -> int:
    cnt=1
    num=nums[0]
    for i in range(1,len(nums)):
        if nums[i]==num:
            cnt+=1
        else:
            cnt-=1
        if cnt==0:
            num=nums[i]
            cnt=1
    return num
```

```python

###Maximum Subarray Problem


class Solution:
    def maxSubArray(self, nums):
        s = 0
        maxi=float("-inf")
        for num in nums:
            s+= num
            maxi=max(maxi,s)
            if s<0:
                s=0
        return maxi
```
