```python




# 189. Rotate Array

import collections

class Solution:
    def rotate(self, nums: list[int], k: int) -> None:
        dq = collections.deque()

        # deque methods:
        # dq.pop()
        # dq.popleft()
        # dq.append()
        # dq.appendleft()

        # Add all elements to deque
        for num in nums:
            dq.append(num)

        # Rotate right k times
        for i in range(k):
            x = dq.pop()
            dq.appendleft(x)

        # Put deque elements back into nums
        n = len(dq)

        for i in range(n):
            nums[i] = dq.popleft()
```

```python

###Optimal solution for Rotate Array

from typing import List

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        #if k is largers we could reduce by taking 
        k = k % len(nums)

        ##Function to reverse a part of array
        def reverse(l,r):
            #Continue the loop until pointers meet
            while l<r:
                ##Swap left and right
                nums[l],nums[r]=nums[r],nums[l]
                ##Move the left pointer one and right pointer -1
                l+=1
                r-=1
        ##Reverse the entire array
        reverse(0,len(nums)-1)
        ##Reverse the array before k
        reverse(0,k-1)
        ##Reverse the array after k-1
        reverse(k,len(nums)-1)
```

```python
###Missing number in an array

from typing import List

class Solution:
    def missingNumber(self, nums: List[int]) -> int:

        n = len(nums)

        originalSum = 0

        for i in range(0, n + 1):
            originalSum += i

        arrSum = 0

        for num in nums:
            arrSum += num

        missing = originalSum - arrSum

        return missing
```
