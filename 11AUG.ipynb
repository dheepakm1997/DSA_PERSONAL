```python

####TO FIND ODD OR EVEN NUMBER###

arr = [5,6,7]
even =[]
odd=[]

for i,num in enumerate(arr):
    if num %2 !=0:
        odd.append(num)
    else:
        even.append(num)

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
##TO FIND POSITIVE OR NEGATIVE NUMBER###

arr=[5,6,7,8,-1-2,-4,6,-7,8,9,10]
positive=[]
negative=[]

for i,num in enumerate(arr):
    if num >=0:
        positive.append(num)
    else:
        negative.append(num)
print("Positive numbers:", positive)
print("Negative numbers:", negative)
```

```python
## To get the zeroes at the end of the list###
arr=[5,6,7,8,-1-2,-4,0,0,0,6,-7,8,9,10]
lst=[]
count=0
for i,num in enumerate(arr):
    if num==0:
        count+=1
    else:
        lst.append(num)

for i in range(count):
    lst.append(0)
print(lst)
    
```

```python
####TO FIND ROTATED SORTED LIST OR NOT###-Leetcode 1752

from typing import List

nums = [3,4,5,1,2]

class Solution:
    def check(self, nums: List[int]) -> bool:
        n = len(nums)
        arr = sorted(nums)

        for i in range(n):
            x = arr.pop()
            arr.insert(0, x)

            if nums == arr:
                return True

        return False

sol = Solution()
print(sol.check(nums))
```

```python
####Find second largest element in the list###-Leetcode 

class Solution:
    def secondLargestElement(self, nums):
        maxi = float('-inf')
        secMax = float('-inf')

        for i, num in enumerate(nums):
            if num > maxi:
                secMax = maxi
                maxi = num

            elif num > secMax and num != maxi:
                secMax = num

        if secMax == float('-inf'):
            return -1

        return secMax


        
```
